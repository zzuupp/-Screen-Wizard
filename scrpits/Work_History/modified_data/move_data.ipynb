```python
# 필요한 라이브러리 가져오기.
import shutil
import os
import numpy as np
```

#### after 안에 존재하는 메인 이외 파일 복제.
* 현 파일들은 메인 페이지 이외 따로 결측치등을 검사하지 않음.
* 작업시 수정된 파일들을 다시 업로드할 것.

```python
# 원본 디렉토리 (현재 내가 옮기고 싶어하는 파일들이 위치한 dir)
current_local = '/Users/qutejin/Documents/ML_Project/Movie_Analyze/Data/raw/after'
target_local = '/Users/qutejin/Documents/ML_Project/Movie_Analyze/Data/processed/after'


for idx, file in enumerate(os.listdir(current_local)[2:]):
    
    # [2:]로  필요하지 않은 데이터를 걸러냈으므로 바로 옮기자.
    shutil.copy(os.path.join(current_local, file), target_local)
    print(f'{idx} 파일 복사완료 ')
    
```
